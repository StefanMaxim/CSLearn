mask = array of dtype bool used to select or ignore values in an array.
array[~mask] is just the inverse of mask, or the copy of the array with the values swapped

array > x returns a mask, or a 1d a

array[mask] returns a 1d array with only the elements where the mask is true

```python
#1 array creation

a = np.array(object=[1,2,3,4,5])
mask = a > 3
b = np.ma.masked_where(a>3,a**2,0)
a[mask] = 4
a[~mask] = 1
print(b,a)

```
