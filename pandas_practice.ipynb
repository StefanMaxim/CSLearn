```python
#inspection:

#first lets create a random dataframe that we can start operating on.
import torch
import pandas as pd
import numpy as np

#yes, numpy can be multiple dimensions

data = {
    "X-Value": torch.rand(size=(20,)).numpy(force=True),
    "Y-Value": torch.rand(size=(20,)).numpy(force=True),
    "Label": torch.randint(low=1,high=21,size=(20,)).numpy(force=True)
}
my_DF = pd.DataFrame(data=data)
my_DF.head(10)
```

```python
#NOW, THE INSPECTION:

#the first n values
my_DF.head(5)

#the last n values
my_DF.tail(5)

#general info on columns and stuff
my_DF.info()

#returns the size, numrows, numcols
my_DF.shape

#series of dtypes for each column
my_DF.dtypes

#the range of the indexes, from 0 to 20
my_DF.index

#lists the columns
my_DF.columns
```

```python
#Selection + Indexing:
sub_df = my_DF['X-Value']
sub_df2 = my_DF[['X-Value', 'Y-Value']] #NOTE: can pass an index or a list of indexes

print(my_DF.head(5))

#two main ways to slice into a dataframe:
#use df.loc(row_label,col_label)
item = my_DF.loc[1,"X-Value"] #here you are using the actual lables (1 in the name of the index, but can be a or b or something)
#this is set using index=[1,2,3...] when creating the dataframe. and "X-Value is the actual label of teh colums"
print(item)

#you can also use this to index out actual dataframes:
item = my_DF.loc[[1,2]] #this will index out rows 1 and 2 (HERE, YOU ARE PASSING ONE ARG, SO THE OTHER IS ASSUMED TO BE EVERYTHING)
print(item)

#lets suppose i want a series, which is the y-values from 1 to 4:
ser1 = my_DF.loc[1:4,'Y-Value']# you were right, just watch the order
print(f"\n{ser1}")
```

```python
#now, lets use iloc which words directly on indexes
print(my_DF.head(5))

#i want a sub dataframe that consists of X value and Y value for the rows 1 to 3

sub1 = my_DF.iloc[1:3,[0,1]]
print(sub1)

#NOTE: the two brackets are not just syntactic sugar!! it just means that you are passing a list as a parameter
#iloc expects two parameters, which can be lists or 1:2 or stuff like that. THIS IS CRUTIAL.
#IF YOU WANT IT TO SELECT EVERYTHING, USE : , MEANING EVERYTHING
sub2 = my_DF.iloc[:,[1,2]] #meaning give me every row 
```

```python
#cleaning + handling missing data

#isnull returns a boolean dataframe where it checks which is null
isnul = my_DF.isnull()
print(isnul.head(5)) #CAN BE USED WITH DF.sum(axis=0) to sum across columns and return the numner of nulls per column.
#normally, it is in row major order, as is shown for iloc and loc, but here apparantly, axis=0 means sum across columns
#REASON: AXIS=0 MEANS COLLAPSE THE 0TH AXIS, SO IT IS THE SAME AS IN TORCH

#isnull and notnul are just opposites:

#IMPORTANT
my_DF.dropna(axis=0) #this drops rows where there is a null

#this is used for filling in the na values as opposed to dropping them
my_DF.fillna(value=1) #fills all the nas with a single value. 

#can also use the iterpolate to fill in the values
my_DF.interpolate(method='linear')

```
