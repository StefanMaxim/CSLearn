```python
#String Methods:

x = "hello world"

list = x.split(sep=" ",maxsplit=1) #splits string into list of substrings #sep is the separator of the string, by default its whitespace
#maxsplit is the max number of splits

x.strip("o")#removes whitespace by def, or the char specified.

x.join(["a","b","c"])# inserts the given string between each of the other strings

x.lower(),x.upper() #to change case, and a bunch of other .if_somethign methods like is alpha or is upper
```
